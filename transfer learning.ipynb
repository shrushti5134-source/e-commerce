```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
trip=pd.read_csv(r"C:\Users\shrus\Downloads\Trip_advisor_review.csv")
```

```python
import nltk  
from nltk.corpus import stopwords
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer #function
```

```python
nltk.download('vader_lexicon')
```

```python
sent = SentimentIntensityAnalyzer()
```

```python
sent.polarity_scores('i enjoyed watching the movie. Movie was was very good. it was awesome')
```

```python
sent.polarity_scores("today is neutral")
```

```python
#problem statement/ requirement--> comapny has launched a new product
#and user are giving the comments/ feedback --->
#company want to find if overall feedback is +ve or -ve

#CONSIDER rating column is not present
```

```python
score_list=[]
for i in range(0, trip.shape[0]):
    score=sent.polarity_scores(trip['Review'][i])
    score_list.append(score['compound'])
```

```python
score_list
```

```python
import numpy as np
```

```python
avg_score=np.mean(score_list)
print(f"Average:{avg_score}")
```

```python
avg_score=max(score_list)
print(f"Max:{avg_score}")
```

```python
avg_score=min(score_list)
print(f"Min:{avg_score}")
```

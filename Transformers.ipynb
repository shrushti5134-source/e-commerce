```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from transformers import pipeline


summarizer = pipeline("text-generation", model="facebook/bart-large-cnn")

text = """
Natural Language Processing (NLP) is a field of Artificial Intelligence that focuses on the interaction
between computers and humans using natural language. It enables machines to understand, interpret, and generate human language.
it is used in many organizations for various purposes and its  most demanded skill set in data science
"""

summary = summarizer(
    text,
    max_length=50,
    min_length=10
#    do_sample=False
)

print(summary[0]['generated_text'])
```

```python
#Where Transformers are used besides time series
Area	How Transformers are used	Example
📝 NLP / Text	Understand relationships between words	ChatGPT, translation, text classification
🤖 Generative AI / LLMs	Generate and understand human language	GPT, Llama, Claude
🖼️ Computer Vision	Understand relationships between image patches	Vision Transformer (ViT)
🎤 Speech Recognition	Convert speech into text	Speech-to-text systems
🌐 Machine Translation	Translate one language to another	English → Marathi
📄 Document Analysis	Understand long documents	Summarization, question answering
💬 Chatbots	Maintain context across conversation	Customer-support bots
🧬 Biology	Learn patterns in DNA/protein sequences	Protein structure/function prediction
💻 Code Generation	Understand and generate programming code	Code assistants
🎯 Recommendation Systems	Learn user-item interaction sequences	Product/movie recommendations
🔍 Information Retrieval	Understand query and document meaning	Search engines
```

```python
model = pipeline("sentiment-analysis")
```

```python
text = "I really enjoyed this movie. It was amazing!"

result = model(text)

print(result)
```

```python
texts = [
    "I really enjoyed this movie.",
    "This product is terrible.",
    "The course was very useful.",
    "I am disappointed with the service."
]

results = model(texts)

for text, result in zip(texts, results):
    print(text)
    print(result)
    print()
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

result = generator(
    "Artificial intelligence is",
    max_length=50,
    num_return_sequences=1
)

print(result[0]["generated_text"])
```

```python
from transformers import pipeline

classifier = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

text = "This course is very interesting and useful."

result = classifier(text)

print(result)
```

```python
from transformers import pipeline

ner = pipeline(
    "ner",
    aggregation_strategy="simple"
)

text = "Sundar Pichai is the CEO of Google and lives in California."

result = ner(text)

for entity in result:
    print(entity)
```

```python
from transformers import pipeline

qa = pipeline("question-answering")

context = """
Python is a popular programming language used for data analysis,
machine learning, web development and automation.
"""

question = "What is Python used for?"

result = qa(
    question=question,
    context=context
)

print(result)
```

```python
from transformers import pipeline

translator = pipeline(
    "translation_en_to_fr",
    model="Helsinki-NLP/opus-mt-en-fr"
)

text = "I am learning artificial intelligence."

result = translator(text)

print(result[0]["translation_text"])
```

```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

text = "I want to learn Python, SQL and machine learning."

labels = [
    "education",
    "sports",
    "politics",
    "technology",
    "finance"
]

result = classifier(text, labels)

print(result)
```

```python
from transformers import pipeline

fill_mask = pipeline(
    "fill-mask",
    model="bert-base-uncased"
)

text = "Python is a popular [MASK] language."

result = fill_mask(text)

for prediction in result[:5]:
    print(
        prediction["token_str"],
        prediction["score"]
    )
```
